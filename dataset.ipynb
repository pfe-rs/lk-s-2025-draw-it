```python
# OVO SE DRUGO POKRECE

import os
import requests

# Lista kategorija koje želiš da preuzmeš
'''
categories = [
    "The Eiffel Tower",
    "cat",
    "apple",
    # dodaj ostale po potrebi
]
'''

base_url = "https://storage.googleapis.com/quickdraw_dataset/full/simplified/"

save_folder = "quickdraw_simplified_50"
os.makedirs(save_folder, exist_ok=True)

for category in categories:
    clean_name = category  # zadrži razmake za sada
    filename = f"full_simplified_{clean_name}.ndjson"
    url_name = category.replace(" ", "%20")
    url = f"{base_url}{url_name}.ndjson"
    save_path = os.path.join(save_folder, filename)

    print(f"Preuzimam: {url} → {save_path}")
    response = requests.get(url)

    if response.status_code == 200:
        with open(save_path, "wb") as f:
            f.write(response.content)
    else:
        print(f"Neuspešno preuzimanje kategorije: {category} (status {response.status_code})")
```

```python
# OVO SE TRECE PUSTA

import os
import json
import random

input_dir = 'quickdraw_simplified_50'
output_dir = 'balanced_chosen_50'
target_count = 200000 

# Kreiranje izlaznog direktorijuma ako ne postoji
os.makedirs(output_dir, exist_ok=True)

# Prolazak kroz sve NDJSON fajlove
for filename in os.listdir(input_dir):
    if filename.endswith('.ndjson'):
        input_path = os.path.join(input_dir, filename)

        # Čitanje svih linija iz fajla
        with open(input_path, 'r') as f:
            lines = f.readlines()

        print(f"{filename}: {len(lines)} crteža")

        # Ako ih ima više, izaberi nasumično 130000
        if len(lines) > target_count:
            selected_lines = random.sample(lines, target_count)
        else:
            selected_lines = lines  # Sve prepiši ako ih je manje

        output_path = os.path.join(output_dir, filename)

        # Zapisivanje u novi fajl
        with open(output_path, 'w') as f:
            f.writelines(selected_lines)

        print(f"✔ Sačuvano: {filename} ({len(selected_lines)} crteža)")
```

```python
import os, json
from PIL import Image, ImageDraw

input_folder = "balanced_chosen_50"
output_folder = "balanced_chosen_50_images"
os.makedirs(output_folder, exist_ok=True)

def draw_strokes(drawing, size=256):
    img = Image.new("L", (size, size), color=255)
    draw = ImageDraw.Draw(img)
    for stroke in drawing:
        if len(stroke) == 2:
            draw.line(list(zip(*stroke)), fill=0, width=2)
    return img

for fname in os.listdir(input_folder):
    if fname.endswith(".ndjson"):
        cname = fname.replace("full_simplified_", "").replace(".ndjson", "")
        class_dir = os.path.join(output_folder, cname)
        os.makedirs(class_dir, exist_ok=True)

        with open(os.path.join(input_folder, fname)) as f:
            for i, line in enumerate(f):
                output_path = os.path.join(class_dir, f"{cname}_{i}.png")
                if os.path.exists(output_path):
                    continue  # preskoči ako slika već postoji

                drawing = json.loads(line)["drawing"]
                img = draw_strokes(drawing)
                img.save(output_path)
```

```python
top_10 = ["snowman", "potato", "calendar", "mermaid", "banana", "hat", "eyeglasses", "flashlight", "firetruck", "coffee cup"]
```

```python
import os
import shutil
import random

source_folder = "balanced_chosen_50_images"
target_folder = "balanced_chosen_10_images"
target_count = 5000

os.makedirs(target_folder, exist_ok=True)

for class_name in top_10:
    src_dir = os.path.join(source_folder, class_name)
    dst_dir = os.path.join(target_folder, class_name)
    os.makedirs(dst_dir, exist_ok=True)

    if os.path.exists(src_dir):
        all_images = [f for f in os.listdir(src_dir) if f.endswith(".png") or f.endswith(".jpg")]
        print(f"{class_name}: pronađeno {len(all_images)} slika")

        if len(all_images) > target_count:
            selected_images = random.sample(all_images, target_count)
        else:
            selected_images = all_images

        for img_name in selected_images:
            shutil.copy(
                os.path.join(src_dir, img_name),
                os.path.join(dst_dir, img_name)
            )

        print(f"Kopirano: {len(selected_images)} slika iz klase '{class_name}'")
    else:
        print(f"Nedostaje folder: {class_name}")
```
